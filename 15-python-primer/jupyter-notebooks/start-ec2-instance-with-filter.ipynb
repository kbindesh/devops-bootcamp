```python
# This python program starts existing EC2 Instance(s) with a particular tag using its unique ID(s)
# Author: Bindesh K
# Make sure to replace 'your-instance-id' with the actual ID of your EC2 instance 
# References: 
# https://docs.aws.amazon.com/boto3/latest/reference/services/ec2/client/start_instances.html
# https://docs.aws.amazon.com/cli/latest/reference/ec2/describe-instances.html
```

```python
# You must first filter and find the instance ID(s) using describe_instances(), and then pass that ID to start_instances()
```

```python
import boto3

# Initialize the EC2 client
ec2_client = boto3.client('ec2')
```

```python
# Define the tag key and value you want to look for
TAG_KEY = 'Environment'
TAG_VALUE = 'Development'
```

```python
# Find instances with the specific tag
filters = [
    {
        'Name': f'tag:{TAG_KEY}',
        'Values': [TAG_VALUE]
    },
    {
        'Name': 'instance-state-name',
        'Values': ['stopped'] # Only fetch instances that can be started
    }
]
```

```python
response = ec2_client.describe_instances(Filters=filters)
```

```python
# Extract instance IDs from the response
instance_ids = []
for reservation in response['Reservations']:
    for instance in reservation['Instances']:
        instance_ids.append(instance['InstanceId'])
```

```python
if instance_ids:
    print(f"Found {len(instance_ids)} stopped instance(s) matching the tag. Starting now...")
    
    start_response = ec2_client.start_instances(InstanceIds=instance_ids)
    
    # Print the state change for each instance
    for instance in start_response['StartingInstances']:
        print(f"Instance {instance['InstanceId']} is now: {instance['CurrentState']['Name']}")
else:
    print("No stopped instances found matching that specific tag.")
```
