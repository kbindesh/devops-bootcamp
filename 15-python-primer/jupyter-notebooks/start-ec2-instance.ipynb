```python
# This python program starts an existing EC2 Instance using its unique ID
# Author: Bindesh K
# Make sure to replace 'your-instance-id' with the actual ID of your EC2 instance 
# Reference: https://docs.aws.amazon.com/boto3/latest/reference/services/ec2/client/start_instances.html
```

```python
import boto3
```

```python
ec2_client = boto3.client('ec2')
instance_id = 'your-instance-id'  # Replace with your actual instance ID
```

```python
response = ec2_client.start_instances(
    InstanceIds=[instance_id]
)
```

```python
# Extract and print the new state
current_state = response['StartingInstances'][0]['CurrentState']['Name']
print(f"Success! Instance state is now: {current_state}")
```
